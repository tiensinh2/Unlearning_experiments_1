# Load Unlearning Experiments from GitHub

This notebook clones the repository from GitHub and makes the modules available for import, avoiding `ModuleNotFoundError` for `experiment_utils`, `data_loader`, and `unlearning_methods`.

```python
# Clone the repository from GitHub
import subprocess
import sys
import os

REPO_URL = "https://github.com/tiensinh2/Unlearning_experiments_1.git"
REPO_DIR = "Unlearning_experiments_1"

if not os.path.exists(REPO_DIR):
    print(f"Cloning {REPO_URL}...")
    result = subprocess.run([sys.executable, "-m", "pip", "install", "gitpython"], capture_output=True, text=True)
    if result.returncode != 0:
        print(result.stderr)
    
    from git import Repo
    Repo.clone_from(REPO_URL, REPO_DIR)
    print(f"Repository cloned to {REPO_DIR}")
else:
    print(f"Repository already exists at {REPO_DIR}")
    # Optionally pull latest changes
    from git import Repo
    repo = Repo(REPO_DIR)
    origin = repo.remotes.origin
    origin.pull()
    print("Pulled latest changes")
```

```python
# Add the repository to Python path
import sys
import os

REPO_DIR = "Unlearning_experiments_1"
repo_path = os.path.abspath(REPO_DIR)

if repo_path not in sys.path:
    sys.path.insert(0, repo_path)
    print(f"Added {repo_path} to sys.path")
else:
    print(f"{repo_path} already in sys.path")
```
